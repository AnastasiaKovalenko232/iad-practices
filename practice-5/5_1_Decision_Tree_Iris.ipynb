## Split data to Train and Test Sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20) # 20%
```

### Decision Tree - create instance

### Decision Tree - train model

```python
model.fit(X_train, y_train)
```

```python
#version 2.0
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split

# 1. Завантаження даних з відкритого джерела (обходимо заблокований Google Диск)
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
df = pd.read_csv(url)

# Перейменовуємо колонки, щоб вони збігалися з кодом викладача
df.columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']

# 2. Розділення на ознаки та цільову змінну
X = df.drop(columns=['Species'])
y = df['Species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Навчання моделі
model = DecisionTreeClassifier()
model.fit(X_train, y_train)

# 4. Візуалізація дерева рішень
plt.figure(figsize=(12, 10))
plot_tree(model, feature_names=X.columns, class_names=y.unique(), filled=True)
plt.title("Дерево рішень для датасету Iris")
plt.show()
```
