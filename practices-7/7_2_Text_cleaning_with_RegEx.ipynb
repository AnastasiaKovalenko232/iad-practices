# Text cleaning with regular expressions

https://docs.python.org/3.8/library/re.html

RegEx tutorial: https://medium.com/factory-mind/regex-tutorial-a-simple-cheatsheet-by-examples-649dc1c3f285

https://regex101.com

```python
import re
import string
import nltk
from nltk.corpus import stopwords

# === КРОК 1: Тихе завантаження стоп-слів ===
print("Завантажуємо словник стоп-слів...")
nltk.download('stopwords', quiet=True)
stop_words = set(stopwords.words("english"))
print("👍 Словник готовий до роботи!\n")

# === КРОК 2: Оригінальний "брудний" текст із лаби ===
text_dirty = 'Noisy Text Example with a lot of numbers and special symbols: 12345, #abc, @@qwerty, http://xyz. Data cleaning Done'
print(f"Оригінальний текст:\n'{text_dirty}'")
print("-" * 60)

# === КРОК 3: Покрокове очищення через RegEx (як у викладача) ===
# 1. Переведення в нижній регістр
text = text_dirty.lower()

# 2. Видалення згадок користувачів (все, що починається з @)
text = re.sub(r"@\S+", " ", text)

# 3. Видалення веб-посилань (http/https)
text = re.sub(r"https?://\S+", " ", text)

# 4. Видалення хетштегів (#)
text = re.sub(r"#\S+", " ", text)

# 5. Видалення всіх цифр
text = re.sub(r"\d+", " ", text)

# 6. Видалення пунктуації (знаків препинання)
text = text.translate(str.maketrans('', '', string.punctuation))

# 7. Видалення символів переносу рядка та зайвих пробілів
text = re.sub(r'\n', ' ', text)
text = re.sub(r'\s{2,}', ' ', text).strip()

# 8. Видалення англійських стоп-слів (тобто зайвих займенників, прийменників тощо)
cleaned_words = [word for word in text.split() if word not in stop_words]
text_final = ' '.join(cleaned_words)

# === КРОК 4: Виведення кінцевого результату ===
print("Результат після повного очищення:")
print(f"'{text_final}'")

print("\n🚀 БЛОКНОТ 7.2 УСПІШНО ВИКОНАНО!")
```
