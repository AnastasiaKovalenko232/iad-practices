```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import requests
import zipfile
import io
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix

# 1. Безпечне завантаження датасету з репозиторію UCI
url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"
r = requests.get(url)
z = zipfile.ZipFile(io.BytesIO(r.content))
# Зчитуємо файл всередині архіву
df = pd.read_csv(z.open('SMSSpamCollection'), sep='\t', names=['label', 'message'])

# 2. Розбиття на вибірки
X_train, X_test, y_train, y_test = train_test_split(df['message'], df['label'], test_size=0.3, random_state=42)

# 3. Векторизація тексту
vectorizer = CountVectorizer(stop_words='english')
X_train_transformed = vectorizer.fit_transform(X_train)
X_test_transformed = vectorizer.transform(X_test)

# 4. Навчання Наївного Байєса (MultinomialNB)
clf = MultinomialNB(alpha=0.1)
clf.fit(X_train_transformed, y_train)
y_pred = clf.predict(X_test_transformed)

# 5. Метрики оцінки моделі
print("=== Метрики Класифікації ===")
print(classification_report(y_test, y_pred))
print("=== Матриця Помилок ===")
print(confusion_matrix(y_test, y_pred))
```
