```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# generation of synthetic dataset
N = 100
a = 1
b = 3

x = np.random.uniform(0, 5, N)
noise = np.random.normal(0, .5, N)
y = a * x + b + noise

plt.grid()
plt.scatter(x, y, marker = '.', color = 'r')
```

```python
X = x.reshape(-1, 1)
```

```python
x # 1d array
```

```python
X # array where each row is an array to
```

```python
# a * x + b = y

model = LinearRegression()
model.fit(X, y)
model
```

```python
a_found = model.coef_[0]
b_found = model.intercept_
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Генерація даних
x = np.linspace(0, 5, 100)
y = x ** 2 - 3
noise = np.random.normal(0, 0.5, 100)
y_noisy = y + noise

# 2. Побудова графіка
plt.figure(figsize=(8, 5))
plt.grid(True, linestyle='--', alpha=0.6)
plt.scatter(x, y_noisy, marker='.', color='blue', label='Дані з шумом')
plt.plot(x, y, color='red', linewidth=2, label='Ідеальна функція')
plt.title('Кастомна генерація даних')
plt.legend()
plt.show()
```
