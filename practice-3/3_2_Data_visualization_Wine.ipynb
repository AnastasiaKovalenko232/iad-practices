# Data Visualization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets as dss
```

## Load the Data

```python
wine = dss.load_wine()
```

```python
type(wine)
```

```python
wine['feature_names']
```

```python
feature_names = wine['feature_names'][:5]
feature_names
```

```python
wine['target']
```

```python
# create data from the first 5 columns
df = pd.DataFrame(
    data = wine['data'][:, :5],
    columns = feature_names
)

df.info()
```

```python
# add a column
df['target'] = wine['target']
df.info()
```

# Pairwise correlation of columns

```python
corr = df.corr()
fig, ax = plt.subplots(figsize=(5, 4)) # size of the box
sns.heatmap(corr,  cmap='coolwarm', annot=True, fmt=".2f")
plt.show()
```

## Boxplots
for more examples see https://seaborn.pydata.org/generated/seaborn.boxplot.html

```python
for i in range(len(feature_names)):
    figure = plt.figure()
    ax = sns.boxplot(x='target', y=feature_names[i], data=df)
```

## Pairplot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Повністю вимикаємо рожеві попередження розробників Seaborn для чистоти аутпуту
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)

# 1. Завантаження даних безпосередньо з вашого файлу wine.csv
df = pd.read_csv('wine.csv')

# Виділяємо назви ознак (всі колонки, крім цільової 'target')
feature_names = [col for col in df.columns if col != 'target']

print(f"Успішно завантажено {df.shape[0]} рядків датасету Wine.")
print(f"Знайдено ознак для аналізу: {len(feature_names)}")

# 2. Побудова матриці кореляції (Теплокарта)
corr = df.corr()
plt.figure(figsize=(12, 9))
sns.heatmap(corr, cmap='coolwarm', annot=True, fmt=".2f", linewidths=0.5)
plt.title('Матриця кореляції хімічних ознак вина', fontsize=14)
plt.tight_layout()
plt.show()

# 3. ПРАВИЛЬНИЙ ЦИКЛ ДЛЯ БОКСПЛОТІВ (без помилок і попереджень)
# Додано параметри hue та legend=False, як вимагає нова версія Seaborn
for i in range(len(feature_names)):
    plt.figure(figsize=(6, 4))
    sns.boxplot(
        x='target',
        y=feature_names[i],
        data=df,
        hue='target',       # Чітко вказуємо групування для кольору
        palette='Set3',     # Гарна палітра кольорів
        legend=False        # Прибираємо зайве віконце легенди, бо підписи є на осі X
    )
    plt.title(f'Розподіл ознаки "{feature_names[i]}" за класами вин', fontsize=11)
    plt.xlabel('Клас вина (Target)')
    plt.ylabel(feature_names[i])
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# 4. Побудова Pairplot (Взаємозв'язок головних ознак)
# Беремо перші 4 ознаки + target, щоб графік не перевантажував оперативну пам'ять
selected_features = feature_names[:4] + ['target']
sns.pairplot(df[selected_features], hue='target', palette='Set1')
plt.suptitle('Взаємозв\'язок перших чотирьох ознак вина (Pairplot)', y=1.02, fontsize=12)
plt.show()
```
