## Titanic dataset Visualisation with Seaborn

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
url = 'https://drive.google.com/uc?id=1zyYJMa75UXLjmODPRnl4_UrLJ0JKmEOC'
df = pd.read_csv(url)
```

### Missing Data

```python
sns.heatmap(df.isnull(), cmap='viridis')
```

### Survived count

```python
# sns.set_style('whitegrid')
sns.countplot(x='Survived', data=df)
```

### Survived by sex

```python
sns.countplot(x='Survived',hue='Sex',data=df)
```

### Survived by Cab Class

```python
sns.countplot(x='Survived',hue='Pclass',data=df)
```

### Distribution of age

```python
sns.histplot(df['Age'])
```

```python
### Class / Age distribution
```

```python
sns.boxplot(
    x='Pclass',
    y='Age',
    data=df
)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Завантаження даних (локально або через url з вашого файлу)
url = 'https://drive.google.com/uc?id=1zyYJMa75UXLjmODPRnl4_UrLJ0JKmEOC'
df = pd.read_csv(url)

# 2. Графік кількості тих, хто вижив, у розрізі класів кают (Pclass)
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', hue='Pclass', data=df)
plt.title('Кількість виживших (0 - ні, 1 - так) за класами кают')
plt.show()

# 3. Розподіл віку пасажирів (Гістограма з лінією щільності)
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, bins=30, color='purple')
plt.title('Розподіл віку пасажирів Титаніка')
plt.xlabel('Вік')
plt.ylabel('Кількість')
plt.show()

# Додаткова крута візуалізація: Стать проти Виживання
plt.figure(figsize=(6, 4))
sns.countplot(x='Survived', hue='Sex', data=df, palette='Set2')
plt.title('Розподіл виживання за статтю')
plt.show()
```
